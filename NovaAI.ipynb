```python
%%writefile /content/pipeline/audio_pipeline/pipeline/standardize.py

import hashlib
import os
from pathlib import Path
from typing import Generator

import librosa
import numpy as np
import soundfile as sf


def _silence_ratio(audio, threshold):
    return float(np.mean(np.abs(audio) < threshold))


def load_and_resample(path, target_sr):
    audio, sr = librosa.load(path, sr=target_sr, mono=True)
    return audio, sr


def trim_silence(audio, sr, top_db=30):
    trimmed, _ = librosa.effects.trim(audio, top_db=top_db)
    return trimmed


def normalize(audio):
    peak = np.max(np.abs(audio))
    if peak < 1e-8:
        return audio
    return audio / peak


def pad_or_segment(audio, sr, target_duration, hop_duration):
    target_samples = int(target_duration * sr)
    hop_samples    = int(hop_duration * sr)
    n_samples      = len(audio)

    if n_samples <= target_samples:
        padded = np.zeros(target_samples, dtype=np.float32)
        padded[:n_samples] = audio
        yield padded
    else:
        start = 0
        while start + target_samples <= n_samples:
            yield audio[start: start + target_samples].astype(np.float32)
            start += hop_samples
        remaining = audio[start:]
        if len(remaining) >= int(0.5 * sr):
            padded = np.zeros(target_samples, dtype=np.float32)
            padded[:len(remaining)] = remaining
            yield padded


def waveform_hash(audio, dtype='float32'):
    return hashlib.sha256(audio.astype(dtype).tobytes()).hexdigest()


def passes_quality_filter(audio, sr, min_duration, max_silence_ratio, silence_threshold):
    duration = len(audio) / sr
    if duration < min_duration:
        return False, f'too_short ({duration:.2f}s)'
    sil = _silence_ratio(audio, silence_threshold)
    if sil > max_silence_ratio:
        return False, f'mostly_silent ({sil:.2f})'
    return True, ''


def standardize_file(input_path, output_path, target_sr, target_duration,
                     hop_duration, min_duration, max_silence_ratio,
                     silence_threshold, bit_depth='PCM_16'):
    try:
        audio, sr = load_and_resample(input_path, target_sr)
    except Exception:
        return []

    audio = trim_silence(audio, sr)
    audio = normalize(audio)

    ok, _ = passes_quality_filter(audio, sr, min_duration, max_silence_ratio, silence_threshold)
    if not ok:
        return []

    results = []
    base_stem = Path(output_path).stem
    out_dir   = Path(output_path).parent
    out_dir.mkdir(parents=True, exist_ok=True)

    for i, chunk in enumerate(pad_or_segment(audio, sr, target_duration, hop_duration)):
        chunk_path = str(output_path) if i == 0 else \
                     str(out_dir / f'{base_stem}_seg{i:03d}.wav')
        sf.write(chunk_path, chunk, sr, subtype=bit_depth)
        results.append({
            'output_path':   chunk_path,
            'waveform_hash': waveform_hash(chunk),
            'duration':      len(chunk) / sr,
            'sr':            sr,
            'chunk_index':   i,
        })
    return results

print('standardize.py written.')
```

```python
%%writefile /content/pipeline/audio_pipeline/pipeline/deduplicate.py

import numpy as np
import pandas as pd
import librosa

DATASET_PRIORITY = {
    'desed': 0, 'wilddesed': 1, 'sins': 2, 'mimii': 3, 'datasec': 4
}
ANNOTATION_PRIORITY = {'strong': 0, 'weak': 1, 'none': 2}


def _priority_score(row):
    dp = DATASET_PRIORITY.get(str(row.get('source_dataset', '')), 99)
    ap = ANNOTATION_PRIORITY.get(str(row.get('annotation_type', 'none')), 2)
    return dp * 10 + ap


def deduplicate_exact(df):
    df = df.copy()
    df['_priority'] = df.apply(_priority_score, axis=1)
    df = df.sort_values('_priority')
    mask = ~df['waveform_hash'].duplicated(keep='first')
    removed = (~mask).sum()
    df = df[mask].drop(columns=['_priority'])
    print(f'  [Exact dedup] Removed {removed:,} exact duplicates → {len(df):,} clips remaining')
    return df.reset_index(drop=True)


def _compute_embedding(audio_path, sr, n_mels, n_fft, hop_length):
    try:
        y, _ = librosa.load(audio_path, sr=sr, mono=True)
        mel  = librosa.feature.melspectrogram(
            y=y, sr=sr, n_fft=n_fft, hop_length=hop_length, n_mels=n_mels)
        log_mel = librosa.power_to_db(mel + 1e-9)
        emb  = np.concatenate([log_mel.mean(axis=1), log_mel.std(axis=1)])
        norm = np.linalg.norm(emb)
        return emb / norm if norm > 1e-8 else emb
    except Exception:
        return None


def deduplicate_near(df, threshold, sr, n_mels, n_fft, hop_length, batch_size=2000):
    print(f'  [Near dedup]  Computing embeddings for {len(df):,} clips...')
    embeddings, valid = [], []
    for path in df['output_path']:
        emb = _compute_embedding(path, sr, n_mels, n_fft, hop_length)
        embeddings.append(emb if emb is not None else np.zeros(2 * n_mels))
        valid.append(emb is not None)

    E    = np.stack(embeddings).astype(np.float32)
    keep = np.ones(len(df), dtype=bool)

    for label in df['unified_label'].unique():
        idx = np.where(df['unified_label'].values == label)[0]
        if len(idx) < 2:
            continue
        for i in range(0, len(idx), batch_size):
            batch = idx[i: i + batch_size]
            sim   = E[batch] @ E[idx].T
            for bi, gi in enumerate(range(i, min(i + batch_size, len(idx)))):
                if not keep[idx[gi]]:
                    continue
                near = np.where(sim[bi] >= threshold)[0]
                near = near[near != gi]
                for nj in near:
                    gnj = idx[nj]
                    if not keep[gnj]:
                        continue
                    if _priority_score(df.iloc[idx[gi]]) <= _priority_score(df.iloc[gnj]):
                        keep[gnj] = False
                    else:
                        keep[idx[gi]] = False
                        break

    removed = (~keep).sum()
    print(f'  [Near dedup]  Removed {removed:,} near-duplicates → {keep.sum():,} remaining')
    return df[keep].reset_index(drop=True)


def run_deduplication(df, threshold, sr, n_mels, n_fft, hop_length):
    print(f'\nDEDUPLICATION (starting with {len(df):,} clips)')
    df = deduplicate_exact(df)
    df = deduplicate_near(df, threshold, sr, n_mels, n_fft, hop_length)
    print(f'  Final: {len(df):,} clips\n')
    return df

print('deduplicate.py written.')
```

```python
%%writefile /content/pipeline/audio_pipeline/pipeline/split.py

import numpy as np
import pandas as pd
from pathlib import Path


def stratified_split(df, ratios, seed):
    rng = np.random.default_rng(seed)

    if 'recording_id' not in df.columns:
        df = df.copy()
        df['recording_id'] = df['clip_id']

    stratum_key = df['unified_label'].astype(str) + '||' + df['source_dataset'].astype(str)
    split_col   = pd.Series(['train'] * len(df), index=df.index)
    warnings_list = []

    val_ratio  = ratios['val']
    test_ratio = ratios['test']

    for stratum in stratum_key.unique():
        mask   = (stratum_key == stratum)
        idx    = df.index[mask].to_numpy()
        groups = df.loc[idx, 'recording_id'].values
        unique_groups = np.unique(groups)

        if len(unique_groups) < 3:
            warnings_list.append(f"Stratum '{stratum}' has <3 recordings — all assigned to train.")
            continue

        n_test  = max(1, int(round(len(unique_groups) * test_ratio)))
        n_val   = max(1, int(round(len(unique_groups) * val_ratio)))
        n_train = max(1, len(unique_groups) - n_val - n_test)

        perm    = rng.permutation(len(unique_groups))
        shuffled = unique_groups[perm]
        train_g  = set(shuffled[:n_train])
        val_g    = set(shuffled[n_train: n_train + n_val])
        test_g   = set(shuffled[n_train + n_val:])

        for ci, grp in zip(idx, groups):
            if grp in val_g:
                split_col.at[ci] = 'val'
            elif grp in test_g:
                split_col.at[ci] = 'test'

    df = df.copy()
    df['split'] = split_col

    print('\n  Split distribution:')
    for s in ['train', 'val', 'test']:
        n   = (df['split'] == s).sum()
        pct = 100 * n / len(df)
        print(f'    {s:6s}: {n:7,} clips ({pct:.1f}%)')

    return df, warnings_list


def save_split_csvs(df, output_dir):
    meta_dir = Path(output_dir) / 'metadata'
    meta_dir.mkdir(parents=True, exist_ok=True)
    df.to_csv(meta_dir / 'master.csv', index=False)
    print(f'\n  Saved master.csv ({len(df):,} rows)')
    for split in ['train', 'val', 'test']:
        sub = df[df['split'] == split]
        sub.to_csv(meta_dir / f'{split}.csv', index=False)
        print(f'  Saved {split}.csv ({len(sub):,} rows)')

print('split.py written.')
```

```python
%%writefile /content/pipeline/audio_pipeline/pipeline/export_stats.py

import json
from pathlib import Path
import pandas as pd
import sys
sys.path.insert(0, '/content/pipeline/audio_pipeline')
from configs.ontology import CLASS_TO_IDX, UNIFIED_CLASSES


def build_stats(df, warnings):
    stats = {
        'total_clips':      int(len(df)),
        'total_duration_h': round(df['duration'].sum() / 3600, 2),
        'n_classes':        int(df['unified_label'].nunique()),
        'n_datasets':       int(df['source_dataset'].nunique()),
        'splits': {},
        'per_class': {},
        'per_dataset': {},
        'warnings': warnings,
    }
    for split in ['train', 'val', 'test']:
        sub = df[df['split'] == split]
        stats['splits'][split] = {
            'clips':      int(len(sub)),
            'duration_h': round(sub['duration'].sum() / 3600, 2),
        }
    for label, group in df.groupby('unified_label'):
        stats['per_class'][label] = {
            'total': int(len(group)),
            'domain': UNIFIED_CLASSES.get(label, {}).get('domain', 'unknown'),
            'train': int((group['split'] == 'train').sum()),
            'val':   int((group['split'] == 'val').sum()),
            'test':  int((group['split'] == 'test').sum()),
            'sources': group['source_dataset'].value_counts().to_dict(),
        }
    counts = [v['total'] for v in stats['per_class'].values()]
    if counts:
        stats['imbalance_ratio'] = round(max(counts) / max(min(counts), 1), 1)
    for ds, group in df.groupby('source_dataset'):
        stats['per_dataset'][ds] = {
            'clips':      int(len(group)),
            'duration_h': round(group['duration'].sum() / 3600, 2),
            'train':      int((group['split'] == 'train').sum()),
            'val':        int((group['split'] == 'val').sum()),
            'test':       int((group['split'] == 'test').sum()),
        }
    return stats


def save_class_map(output_dir):
    meta_dir = Path(output_dir) / 'metadata'
    meta_dir.mkdir(parents=True, exist_ok=True)
    class_map = {
        'label_to_idx': CLASS_TO_IDX,
        'idx_to_label': {str(v): k for k, v in CLASS_TO_IDX.items()},
        'n_classes': len(CLASS_TO_IDX),
    }
    with open(meta_dir / 'class_map.json', 'w') as f:
        json.dump(class_map, f, indent=2)
    print(f'  Saved class_map.json ({len(CLASS_TO_IDX)} classes)')


def save_stats(stats, output_dir):
    meta_dir = Path(output_dir) / 'metadata'
    with open(meta_dir / 'stats.json', 'w') as f:
        json.dump(stats, f, indent=2)
    print('  Saved stats.json')


def print_report(stats):
    sep = '=' * 56
    print(f'\n{sep}')
    print('  UNIFIED DATASET — FINAL REPORT')
    print(sep)
    print(f"  Total clips    : {stats['total_clips']:,}")
    print(f"  Total duration : {stats['total_duration_h']} hours")
    print(f"  Classes        : {stats['n_classes']}")
    print(f"  Source datasets: {stats['n_datasets']}")
    print(f"  Imbalance ratio: {stats.get('imbalance_ratio', 'N/A')}x")
    print(f"\n  {'Split':<8} {'Clips':>8} {'Hours':>10}")
    print(f"  {'-'*30}")
    for split, s in stats['splits'].items():
        print(f"  {split:<8} {s['clips']:>8,} {s['duration_h']:>10.1f}")
    print(f"\n  {'Dataset':<18} {'Clips':>8} {'Hours':>8}")
    print(f"  {'-'*36}")
    for ds, s in sorted(stats['per_dataset'].items(), key=lambda x: -x[1]['clips']):
        print(f"  {ds:<18} {s['clips']:>8,} {s['duration_h']:>8.1f}")
    print(sep)


def export_all(df, warnings, output_dir):
    print('\nEXPORTING METADATA')
    save_class_map(output_dir)
    stats = build_stats(df, warnings)
    save_stats(stats, output_dir)
    print_report(stats)

print('export_stats.py written.')
```

```python
%%writefile /content/pipeline/audio_pipeline/pipeline/ingest.py

import os
import json
from pathlib import Path
import pandas as pd


def ingest_desed(root):
    records = []
    for tsv_path in (root / 'metadata').rglob('*strong*.tsv'):
        try:
            df = pd.read_csv(tsv_path, sep='\t')
            for _, row in df.iterrows():
                audio_path = root / 'audio' / row['filename']
                records.append({
                    'input_path': str(audio_path),
                    'raw_label': str(row['event_label']),
                    'annotation_type': 'strong',
                    'onset': float(row.get('onset', 0.0)),
                    'offset': float(row.get('offset', 0.0)),
                    'recording_id': Path(row['filename']).stem,
                    'source_dataset': 'desed',
                    'license': 'CC BY 4.0',
                    'snr_db': None,
                })
        except Exception as e:
            print(f'  [DESED] Warning: {e}')

    weak_tsv = root / 'metadata' / 'train' / 'weak.tsv'
    if weak_tsv.exists():
        try:
            df = pd.read_csv(weak_tsv, sep='\t')
            for _, row in df.iterrows():
                audio_path = root / 'audio' / row['filename']
                for lbl in str(row['event_label']).split(','):
                    records.append({
                        'input_path': str(audio_path),
                        'raw_label': lbl.strip(),
                        'annotation_type': 'weak',
                        'onset': None, 'offset': None,
                        'recording_id': Path(row['filename']).stem,
                        'source_dataset': 'desed',
                        'license': 'CC BY 4.0',
                        'snr_db': None,
                    })
        except Exception as e:
            print(f'  [DESED] Warning weak.tsv: {e}')

    print(f'  [DESED]     {len(records):,} annotation entries')
    return records


def ingest_wilddesed(root):
    records = []
    for tsv_path in (root / 'metadata').rglob('*.tsv'):
        try:
            df = pd.read_csv(tsv_path, sep='\t')
            has_strong = 'onset' in df.columns and 'offset' in df.columns
            for _, row in df.iterrows():
                audio_path = root / 'audio' / row['filename']
                records.append({
                    'input_path': str(audio_path),
                    'raw_label': str(row.get('event_label', 'unknown')),
                    'annotation_type': 'strong' if has_strong else 'weak',
                    'onset': float(row['onset']) if has_strong else None,
                    'offset': float(row['offset']) if has_strong else None,
                    'recording_id': Path(row['filename']).stem,
                    'source_dataset': 'wilddesed',
                    'license': 'CC BY 4.0',
                    'snr_db': None,
                })
        except Exception as e:
            print(f'  [WildDESED] Warning: {e}')
    print(f'  [WildDESED] {len(records):,} annotation entries')
    return records


def ingest_sins(root):
    records = []
    annotation_csv = root / 'annotations.csv'
    if not annotation_csv.exists():
        candidates = list(root.rglob('annotations*.csv'))
        annotation_csv = candidates[0] if candidates else None

    if annotation_csv is None:
        print(f'  [SINS] No annotations.csv found')
        return records

    try:
        df = pd.read_csv(annotation_csv)
        for _, row in df.iterrows():
            fname = str(row.get('recording', row.get('filename', '')))
            audio_path = root / 'audio' / fname
            records.append({
                'input_path': str(audio_path),
                'raw_label': str(row.get('activity', row.get('event_label', 'other'))),
                'annotation_type': 'strong',
                'onset': float(row.get('onset', 0.0)),
                'offset': float(row.get('offset', 0.0)),
                'recording_id': Path(fname).stem,
                'source_dataset': 'sins',
                'license': 'CC BY 4.0',
                'snr_db': None,
            })
    except Exception as e:
        print(f'  [SINS] Warning: {e}')
    print(f'  [SINS]      {len(records):,} annotation entries')
    return records


def ingest_mimii(root):
    records = []
    snr_levels = {'-6dB': -6, '0dB': 0, '6dB': 6}
    for snr_str, snr_val in snr_levels.items():
        snr_dir = root / snr_str
        if not snr_dir.exists():
            continue
        for machine in ['valve', 'pump', 'fan', 'slider']:
            machine_dir = snr_dir / machine
            if not machine_dir.exists():
                continue
            for model_dir in machine_dir.iterdir():
                if not model_dir.is_dir():
                    continue
                for condition in ['normal', 'abnormal']:
                    cond_dir = model_dir / condition
                    if not cond_dir.exists():
                        continue
                    for wav in cond_dir.glob('*.wav'):
                        records.append({
                            'input_path': str(wav),
                            'raw_label': f'{machine}/{"anomaly" if condition == "abnormal" else "normal"}',
                            'annotation_type': 'weak',
                            'onset': None, 'offset': None,
                            'recording_id': wav.stem,
                            'source_dataset': 'mimii',
                            'license': 'CC BY-SA 4.0',
                            'snr_db': snr_val,
                        })
    print(f'  [MIMII]     {len(records):,} audio files')
    return records


def ingest_datasec(root):
    records = []
    datasec_dir = root / 'DataSEC'
    if datasec_dir.exists():
        for class_dir in datasec_dir.iterdir():
            if not class_dir.is_dir():
                continue
            for wav in class_dir.glob('*.wav'):
                records.append({
                    'input_path': str(wav),
                    'raw_label': class_dir.name,
                    'annotation_type': 'weak',
                    'onset': None, 'offset': None,
                    'recording_id': wav.stem,
                    'source_dataset': 'datasec',
                    'license': 'CC BY 4.0',
                    'snr_db': None,
                })

    datased_dir = root / 'DataSED'
    if datased_dir.exists():
        ann_dir = datased_dir / 'annotations'
        if ann_dir.exists():
            for csv_path in ann_dir.glob('*.csv'):
                try:
                    df = pd.read_csv(csv_path)
                    rec_name = csv_path.stem
                    for _, row in df.iterrows():
                        audio_path = datased_dir / 'recordings' / f'{rec_name}.wav'
                        records.append({
                            'input_path': str(audio_path),
                            'raw_label': str(row.get('label', 'unknown')),
                            'annotation_type': 'strong',
                            'onset': float(row.get('onset', 0.0)),
                            'offset': float(row.get('offset', 0.0)),
                            'recording_id': rec_name,
                            'source_dataset': 'datasec',
                            'license': 'CC BY 4.0',
                            'snr_db': None,
                        })
                except Exception as e:
                    print(f'  [DataSEC] Warning: {e}')

    print(f'  [DataSEC]   {len(records):,} annotation entries')
    return records


def scan_dataset(name, root_dir, allowed_fsd_licenses=None):
    root = Path(root_dir)
    if not root.exists():
        print(f'  [SKIP] {name}: not found at {root}')
        return []
    dispatch = {
        'desed':     lambda: ingest_desed(root),
        'wilddesed': lambda: ingest_wilddesed(root),
        'sins':      lambda: ingest_sins(root),
        'mimii':     lambda: ingest_mimii(root),
        'datasec':   lambda: ingest_datasec(root),
    }
    fn = dispatch.get(name)
    if fn is None:
        print(f'  [SKIP] No adapter for {name}')
        return []
    return fn()

print('ingest.py written.')
```

```python
%%writefile /content/pipeline/audio_pipeline/scripts/dataset_loader.py

import json, random
from pathlib import Path
import librosa
import numpy as np
import pandas as pd

try:
    import torch
    from torch.utils.data import Dataset
    TORCH_AVAILABLE = True
except ImportError:
    class Dataset: pass
    TORCH_AVAILABLE = False

TARGET_SR  = 16_000
DURATION   = 10.0
N_FFT      = 1024
HOP_LENGTH = 512
N_MELS     = 128
F_MIN      = 20
F_MAX      = 8_000

ANOMALY_LABELS = {
    'machine_valve_anomalous', 'machine_pump_anomalous',
    'machine_fan_anomalous',   'machine_slide_rail_anomalous',
}


class UnifiedAudioDataset(Dataset):
    def __init__(self, csv_path, audio_root, class_map,
                 task='classification', augment=False, cache_mels=False):
        assert task in ('classification', 'sed', 'anomaly')
        self.df         = pd.read_csv(csv_path)
        self.audio_root = Path(audio_root)
        self.task       = task
        self.augment    = augment
        self.cache_mels = cache_mels
        self._cache     = {}

        if task == 'anomaly':
            self.df = self.df[self.df['unified_label'].str.startswith('machine_')]
        elif task == 'sed':
            self.df = self.df[self.df['annotation_type'] == 'strong']
        self.df = self.df.reset_index(drop=True)

        with open(class_map) as f:
            cm = json.load(f)
        self.label_to_idx = cm['label_to_idx']
        self.idx_to_label = {int(k): v for k, v in cm['idx_to_label'].items()}
        self.n_classes    = cm['n_classes']

    def __len__(self):
        return len(self.df)

    def _load_audio(self, path):
        y, _ = librosa.load(path, sr=TARGET_SR, mono=True, duration=DURATION)
        target = int(TARGET_SR * DURATION)
        if len(y) < target:
            y = np.pad(y, (0, target - len(y)))
        return y[:target].astype(np.float32)

    def _log_mel(self, y):
        mel = librosa.feature.melspectrogram(
            y=y, sr=TARGET_SR, n_fft=N_FFT, hop_length=HOP_LENGTH,
            n_mels=N_MELS, fmin=F_MIN, fmax=F_MAX, power=2.0)
        log_mel = librosa.power_to_db(mel + 1e-9, ref=np.max)
        log_mel = (log_mel - log_mel.min()) / (log_mel.max() - log_mel.min() + 1e-8)
        return log_mel.astype(np.float32)

    def _augment(self, y):
        shift = random.randint(-int(0.10 * len(y)), int(0.10 * len(y)))
        y = np.roll(y, shift)
        gain_db = random.uniform(-3.0, 3.0)
        return np.clip(y * (10 ** (gain_db / 20.0)), -1.0, 1.0)

    def __getitem__(self, idx):
        row  = self.df.iloc[idx]
        path = str(row.get('output_path', ''))
        if not Path(path).exists():
            label = str(row['unified_label']).replace('/', '_')
            fname = Path(path).name if path else f"{row.get('clip_id', idx)}.wav"
            path  = str(self.audio_root / label / fname)

        if self.cache_mels and path in self._cache:
            log_mel = self._cache[path]
        else:
            y = self._load_audio(path)
            if self.augment:
                y = self._augment(y)
            log_mel = self._log_mel(y)
            if self.cache_mels:
                self._cache[path] = log_mel

        spec = torch.from_numpy(log_mel).unsqueeze(0) if TORCH_AVAILABLE else log_mel[np.newaxis]

        if self.task == 'classification':
            ci = self.label_to_idx.get(str(row['unified_label']), 0)
            return spec, (torch.tensor(ci, dtype=torch.long) if TORCH_AVAILABLE else ci)

        elif self.task == 'sed':
            ci     = self.label_to_idx.get(str(row['unified_label']), 0)
            onset  = float(row.get('onset',  0.0) or 0.0)
            offset = float(row.get('offset', DURATION) or DURATION)
            if TORCH_AVAILABLE:
                return spec, {
                    'class_idx': torch.tensor(ci, dtype=torch.long),
                    'onset':     torch.tensor(onset, dtype=torch.float32),
                    'offset':    torch.tensor(offset, dtype=torch.float32),
                }
            return spec, {'class_idx': ci, 'onset': onset, 'offset': offset}

        elif self.task == 'anomaly':
            is_anom = int(str(row['unified_label']) in ANOMALY_LABELS)
            return spec, (torch.tensor(is_anom, dtype=torch.float32) if TORCH_AVAILABLE else is_anom)

    def class_weights(self):
        counts  = self.df['unified_label'].value_counts()
        weights = np.zeros(self.n_classes, dtype=np.float32)
        for label, count in counts.items():
            i = self.label_to_idx.get(label)
            if i is not None:
                weights[i] = 1.0 / (count + 1e-6)
        weights = weights / weights.sum() * self.n_classes
        return torch.from_numpy(weights) if TORCH_AVAILABLE else weights

    def __repr__(self):
        return f"UnifiedAudioDataset(n={len(self)}, task='{self.task}', n_classes={self.n_classes})"

print('dataset_loader.py written.')
```

```python
%%writefile /content/pipeline/audio_pipeline/build_dataset.py

import os, sys, uuid, time, argparse
from concurrent.futures import ProcessPoolExecutor, as_completed
from pathlib import Path

import pandas as pd
from tqdm import tqdm

sys.path.insert(0, '/content/pipeline/audio_pipeline')

from configs.settings  import (RAW_DATA_DIR, OUTPUT_DIR, TARGET_SR,
                                TARGET_DURATION, HOP_DURATION, N_FFT,
                                HOP_LENGTH, N_MELS, DEDUP_THRESHOLD,
                                SPLIT_RATIOS, RANDOM_SEED, MIN_CLIP_DURATION,
                                MAX_SILENCE_RATIO, SILENCE_THRESHOLD,
                                FSD50K_ALLOWED_LICENCES, DATASET_DIRS,
                                AUDIO_FORMAT, BIT_DEPTH)
from configs.ontology  import LABEL_MAP
from pipeline.ingest   import scan_dataset
from pipeline.standardize import standardize_file
from pipeline.split    import stratified_split, save_split_csvs
from pipeline.export_stats import export_all


def parse_args():
    p = argparse.ArgumentParser()
    p.add_argument('--dry-run',    action='store_true')
    p.add_argument('--skip-dedup', action='store_true')
    p.add_argument('--dataset',    type=str, default=None)
    p.add_argument('--workers',    type=int, default=2)
    return p.parse_args()


def step_ingest(target_dataset):
    print(f'\n{"="*50}\nINGEST & LABEL MAPPING\n{"="*50}')
    all_records = []
    datasets = DATASET_DIRS if target_dataset is None else {target_dataset: DATASET_DIRS[target_dataset]}
    for ds_name, ds_folder in datasets.items():
        raw_path = os.path.join(RAW_DATA_DIR, ds_folder)
        records  = scan_dataset(ds_name, raw_path, FSD50K_ALLOWED_LICENCES)
        mapped = skipped = 0
        for r in records:
            unified = LABEL_MAP.get(r['raw_label']) or \
                      LABEL_MAP.get(r['raw_label'].strip().replace(' ', '_'))
            if unified:
                r['unified_label'] = unified
                mapped += 1
            else:
                r['unified_label'] = None
                skipped += 1
        records = [r for r in records if r['unified_label'] is not None]
        print(f'  {ds_name:<12}: {mapped:,} mapped | {skipped:,} skipped')
        all_records.extend(records)
    print(f'\n  Total: {len(all_records):,} entries')
    return all_records


def _worker(args):
    record, output_base = args
    input_path = record['input_path']
    if not os.path.exists(input_path):
        return []
    ds    = record['source_dataset']
    label = record['unified_label'].replace('/', '_')
    stem  = Path(input_path).stem
    out_path = os.path.join(output_base, ds, label, f'{stem}.wav')
    os.makedirs(os.path.dirname(out_path), exist_ok=True)
    chunks = standardize_file(
        input_path, out_path, TARGET_SR, TARGET_DURATION, HOP_DURATION,
        MIN_CLIP_DURATION, MAX_SILENCE_RATIO, SILENCE_THRESHOLD, BIT_DEPTH)
    results = []
    for chunk in chunks:
        r = dict(record)
        r.update(chunk)
        r['clip_id'] = str(uuid.uuid4())
        results.append(r)
    return results


def step_standardize(records, workers):
    print(f'\n{"="*50}\nSTANDARDIZE AUDIO\n{"="*50}')
    unique_inputs = {}
    for r in records:
        key = (r['input_path'], r['unified_label'])
        if key not in unique_inputs:
            unique_inputs[key] = r

    audio_dir = os.path.join(OUTPUT_DIR, 'audio', 'all')
    args_list = [(r, audio_dir) for r in unique_inputs.values()]
    print(f'  Processing {len(args_list):,} files with {workers} workers...')

    all_chunks = []
    with ProcessPoolExecutor(max_workers=workers) as ex:
        futs = {ex.submit(_worker, a): a for a in args_list}
        for fut in tqdm(as_completed(futs), total=len(futs),
                        desc='  Standardizing', ncols=70):
            try:
                all_chunks.extend(fut.result())
            except Exception:
                pass

    df = pd.DataFrame(all_chunks)
    print(f'\n  Produced {len(df):,} clips')
    return df


def step_split(df):
    print(f'\n{"="*50}\nSTRATIFIED SPLIT\n{"="*50}')
    return stratified_split(df, SPLIT_RATIOS, RANDOM_SEED)


def step_reorganize(df):
    import shutil
    print(f'\n{"="*50}\nREORGANIZE INTO SPLIT FOLDERS\n{"="*50}')
    new_paths = []
    for _, row in tqdm(df.iterrows(), total=len(df), desc='  Moving', ncols=70):
        src = row['output_path']
        if not os.path.exists(src):
            new_paths.append(src)
            continue
        split    = row['split']
        label    = row['unified_label'].replace('/', '_')
        fname    = Path(src).name
        dest_dir = Path(OUTPUT_DIR) / 'audio' / split / label
        dest_dir.mkdir(parents=True, exist_ok=True)
        dest = dest_dir / fname
        if not dest.exists():
            shutil.move(src, dest)
        new_paths.append(str(dest))
    df = df.copy()
    df['output_path'] = new_paths
    staging = Path(OUTPUT_DIR) / 'audio' / 'all'
    if staging.exists():
        shutil.rmtree(staging, ignore_errors=True)
    print('  Done.')
    return df


def main():
    args = parse_args()
    t0 = time.time()

    # Load existing master.csv if it exists (for incremental runs)
    master_csv = os.path.join(OUTPUT_DIR, 'metadata', 'master.csv')
    existing_df = pd.read_csv(master_csv) if os.path.exists(master_csv) else pd.DataFrame()

    records = step_ingest(args.dataset)
    if not records:
        print('No records found.')
        return

    df = step_standardize(records, args.workers)
    if df.empty:
        print('No clips survived standardization.')
        return

    # Append to existing clips from previous dataset runs
    if not existing_df.empty:
        df = pd.concat([existing_df, df], ignore_index=True)
        print(f'\n  Combined with existing: {len(df):,} total clips')

    df, warnings = step_split(df)
    df = step_reorganize(df)

    save_split_csvs(df, OUTPUT_DIR)
    export_all(df, warnings, OUTPUT_DIR)

    print(f'\n  Done in {(time.time()-t0)/60:.1f} minutes.')


if __name__ == '__main__':
    main()

print('build_dataset.py written.')
```

```python
import os, sys

# Create folders
for folder in [
    '/content/pipeline/audio_pipeline/configs',
    '/content/pipeline/audio_pipeline/pipeline',
    '/content/pipeline/audio_pipeline/scripts',
]:
    os.makedirs(folder, exist_ok=True)

for pkg in [
    '/content/pipeline/audio_pipeline',
    '/content/pipeline/audio_pipeline/configs',
    '/content/pipeline/audio_pipeline/pipeline',
    '/content/pipeline/audio_pipeline/scripts',
]:
    open(os.path.join(pkg, '__init__.py'), 'w').close()

sys.path.insert(0, '/content/pipeline/audio_pipeline')

# ── configs/ontology.py ───────────────────────────────────────────────────
open('/content/pipeline/audio_pipeline/configs/ontology.py', 'w').write('''
UNIFIED_CLASSES = {
    "alarm_bell_ringing":           {"domain": "domestic", "task": ["classification", "sed"]},
    "blender":                      {"domain": "domestic", "task": ["classification", "sed"]},
    "cat":                          {"domain": "domestic", "task": ["classification", "sed"]},
    "dog":                          {"domain": "domestic", "task": ["classification", "sed"]},
    "dishes":                       {"domain": "domestic", "task": ["classification", "sed"]},
    "electric_shaver_toothbrush":   {"domain": "domestic", "task": ["classification", "sed"]},
    "frying":                       {"domain": "domestic", "task": ["classification", "sed"]},
    "running_water":                {"domain": "domestic", "task": ["classification", "sed"]},
    "speech":                       {"domain": "domestic", "task": ["classification", "sed"]},
    "vacuum_cleaner":               {"domain": "domestic", "task": ["classification", "sed"]},
    "glass_break":                  {"domain": "domestic", "task": ["classification", "sed"]},
    "door_knock_slam":              {"domain": "domestic", "task": ["classification", "sed"]},
    "telephone_ringtone":           {"domain": "domestic", "task": ["classification", "sed"]},
    "washing_machine":              {"domain": "domestic", "task": ["classification", "sed"]},
    "microwave":                    {"domain": "domestic", "task": ["classification", "sed"]},
    "television":                   {"domain": "domestic", "task": ["classification", "sed"]},
    "footsteps":                    {"domain": "domestic", "task": ["classification", "sed"]},
    "laughter":                     {"domain": "domestic", "task": ["classification", "sed"]},
    "crying":                       {"domain": "domestic", "task": ["classification", "sed"]},
    "cough_sneeze":                 {"domain": "domestic", "task": ["classification", "sed"]},
    "keyboard_mouse":               {"domain": "domestic", "task": ["classification", "sed"]},
    "clock":                        {"domain": "domestic", "task": ["classification", "sed"]},
    "music_indoor":                 {"domain": "domestic", "task": ["classification", "sed"]},
    "activity_cooking":             {"domain": "domestic", "task": ["sed"]},
    "activity_watching_tv":         {"domain": "domestic", "task": ["sed"]},
    "activity_computer_work":       {"domain": "domestic", "task": ["sed"]},
    "activity_showering":           {"domain": "domestic", "task": ["sed"]},
    "activity_toilet":              {"domain": "domestic", "task": ["sed"]},
    "activity_vacuum_cleaning":     {"domain": "domestic", "task": ["sed"]},
    "activity_eating":              {"domain": "domestic", "task": ["sed"]},
    "activity_personal_hygiene":    {"domain": "domestic", "task": ["sed"]},
    "activity_absence":             {"domain": "domestic", "task": ["sed"]},
    "activity_other":               {"domain": "domestic", "task": ["sed"]},
    "machine_valve_normal":         {"domain": "anomaly", "task": ["anomaly"]},
    "machine_valve_anomalous":      {"domain": "anomaly", "task": ["anomaly"]},
    "machine_pump_normal":          {"domain": "anomaly", "task": ["anomaly"]},
    "machine_pump_anomalous":       {"domain": "anomaly", "task": ["anomaly"]},
    "machine_fan_normal":           {"domain": "anomaly", "task": ["anomaly"]},
    "machine_fan_anomalous":        {"domain": "anomaly", "task": ["anomaly"]},
    "machine_slide_rail_normal":    {"domain": "anomaly", "task": ["anomaly"]},
    "machine_slide_rail_anomalous": {"domain": "anomaly", "task": ["anomaly"]},
    "bells":                        {"domain": "outdoor", "task": ["classification"]},
    "birds":                        {"domain": "outdoor", "task": ["classification"]},
    "cat_outdoor":                  {"domain": "outdoor", "task": ["classification"]},
    "chicken_coop":                 {"domain": "outdoor", "task": ["classification"]},
    "cicadas_crickets":             {"domain": "outdoor", "task": ["classification"]},
    "crows_seagulls_magpies":       {"domain": "outdoor", "task": ["classification"]},
    "dog_outdoor":                  {"domain": "outdoor", "task": ["classification"]},
    "glass_break_outdoor":          {"domain": "outdoor", "task": ["classification"]},
    "horn":                         {"domain": "outdoor", "task": ["classification"]},
    "jet_aircraft":                 {"domain": "outdoor", "task": ["classification"]},
    "lawn_mower_brush_cutter":      {"domain": "outdoor", "task": ["classification"]},
    "music_outdoor":                {"domain": "outdoor", "task": ["classification"]},
    "propeller_aircraft_helicopter":{"domain": "outdoor", "task": ["classification"]},
    "siren_alarm_outdoor":          {"domain": "outdoor", "task": ["classification"]},
    "thunder_fireworks_gunshot":    {"domain": "outdoor", "task": ["classification"]},
    "train":                        {"domain": "outdoor", "task": ["classification"]},
    "vacuum_fan_hairdryer_outdoor": {"domain": "outdoor", "task": ["classification"]},
    "vehicle_idling":               {"domain": "outdoor", "task": ["classification"]},
    "vehicle_passby":               {"domain": "outdoor", "task": ["classification"]},
    "voices_outdoor":               {"domain": "outdoor", "task": ["classification"]},
    "wind_turbine":                 {"domain": "outdoor", "task": ["classification"]},
    "workshop_sounds":              {"domain": "outdoor", "task": ["classification"]},
}

LABEL_MAP = {
    "Alarm_bell_ringing": "alarm_bell_ringing",
    "alarm_bell_ringing": "alarm_bell_ringing",
    "Blender": "blender", "blender": "blender",
    "Cat": "cat", "cat": "cat",
    "Dog": "dog", "dog": "dog",
    "Dishes": "dishes", "dishes": "dishes",
    "Electric_shaver_toothbrush": "electric_shaver_toothbrush",
    "electric_shaver_toothbrush": "electric_shaver_toothbrush",
    "Frying": "frying", "frying": "frying",
    "Running_water": "running_water", "running_water": "running_water",
    "Speech": "speech", "speech": "speech",
    "Vacuum_cleaner": "vacuum_cleaner", "vacuum_cleaner": "vacuum_cleaner",
    "Alarm": "alarm_bell_ringing",
    "Smoke_detector,_smoke_alarm": "alarm_bell_ringing",
    "Bell": "alarm_bell_ringing",
    "Glass": "glass_break",
    "Dishes,_pots,_and_pans": "dishes",
    "Water_tap,_faucet": "running_water",
    "Toilet_flush": "running_water",
    "Washing_machine": "washing_machine",
    "Microwave_oven": "microwave",
    "Television": "television",
    "Clock": "clock",
    "Telephone": "telephone_ringtone",
    "Ringtone": "telephone_ringtone",
    "Computer_keyboard": "keyboard_mouse",
    "Mouse": "keyboard_mouse",
    "Door": "door_knock_slam",
    "Knock": "door_knock_slam",
    "Laughter": "laughter",
    "Crying,_sobbing": "crying",
    "Cough": "cough_sneeze",
    "Sneeze": "cough_sneeze",
    "Footsteps": "footsteps",
    "Music": "music_indoor",
    "cooking": "activity_cooking",
    "watching_tv": "activity_watching_tv",
    "computer_work": "activity_computer_work",
    "showering": "activity_showering",
    "toilet": "activity_toilet",
    "vacuum_cleaning": "activity_vacuum_cleaning",
    "eating": "activity_eating",
    "personal_hygiene": "activity_personal_hygiene",
    "absence": "activity_absence",
    "other": "activity_other",
    "valve/normal": "machine_valve_normal",
    "valve/anomaly": "machine_valve_anomalous",
    "pump/normal": "machine_pump_normal",
    "pump/anomaly": "machine_pump_anomalous",
    "fan/normal": "machine_fan_normal",
    "fan/anomaly": "machine_fan_anomalous",
    "slider/normal": "machine_slide_rail_normal",
    "slider/anomaly": "machine_slide_rail_anomalous",
    "Bells": "bells",
    "Birds": "birds",
    "Cat fights and moans": "cat_outdoor",
    "Chicken coop": "chicken_coop",
    "Cicadas and crickets": "cicadas_crickets",
    "Crows seagulls and magpies": "crows_seagulls_magpies",
    "Dog barkings and howlings": "dog_outdoor",
    "Glass breaking": "glass_break_outdoor",
    "Horn": "horn",
    "Jet aircrafts": "jet_aircraft",
    "Lawn mower brush cutter and olive shaker": "lawn_mower_brush_cutter",
    "Propeller aircrafts": "propeller_aircraft_helicopter",
    "Sirens and alarms": "siren_alarm_outdoor",
    "Thunder fireworks and gunshot": "thunder_fireworks_gunshot",
    "Train": "train",
    "Vacuum cleaner fan and hairdryer": "vacuum_fan_hairdryer_outdoor",
    "Vehicle idling": "vehicle_idling",
    "Vehicle pass-by": "vehicle_passby",
    "Voices": "voices_outdoor",
    "Wind turbine": "wind_turbine",
    "Workshop": "workshop_sounds",
}

CLASS_TO_IDX = {cls: i for i, cls in enumerate(sorted(UNIFIED_CLASSES.keys()))}
IDX_TO_CLASS = {i: cls for cls, i in CLASS_TO_IDX.items()}
''')

# ── configs/settings.py ───────────────────────────────────────────────────
open('/content/pipeline/audio_pipeline/configs/settings.py', 'w').write('''
import os
BASE_DIR        = "/content"
RAW_DATA_DIR    = os.path.join(BASE_DIR, "raw_data")
OUTPUT_DIR      = os.path.join(BASE_DIR, "unified_dataset")
CACHE_DIR       = os.path.join(BASE_DIR, ".cache")
TARGET_SR        = 16_000
TARGET_DURATION  = 10.0
HOP_DURATION     = 5.0
TARGET_CHANNELS  = 1
AUDIO_FORMAT     = "wav"
BIT_DEPTH        = "PCM_16"
N_FFT            = 1024
HOP_LENGTH       = 512
N_MELS           = 128
F_MIN            = 20
F_MAX            = 8_000
POWER            = 2.0
HASH_DTYPE       = "float32"
DEDUP_THRESHOLD  = 0.95
EMBED_DIM        = 128
SPLIT_RATIOS     = {"train": 0.70, "val": 0.15, "test": 0.15}
RANDOM_SEED      = 42
STRATIFY_BY      = ["unified_label", "source_dataset"]
MIN_CLIP_DURATION   = 0.5
MAX_SILENCE_RATIO   = 0.90
SILENCE_THRESHOLD   = 0.01
FSD50K_ALLOWED_LICENCES = {
    "Creative Commons Attribution 4.0",
    "Creative Commons Attribution 3.0",
    "Creative Commons 0",
}
DATASET_DIRS = {
    "desed":     "DESED",
    "sins":      "SINS",
    "mimii":     "MIMII",
    "datasec":   "DataSEC_DataSED",
}
''')

# ── pipeline/standardize.py ───────────────────────────────────────────────
open('/content/pipeline/audio_pipeline/pipeline/standardize.py', 'w').write('''
import hashlib, os
from pathlib import Path
import librosa
import numpy as np
import soundfile as sf

def _silence_ratio(audio, threshold):
    return float(np.mean(np.abs(audio) < threshold))

def load_and_resample(path, target_sr):
    audio, sr = librosa.load(path, sr=target_sr, mono=True)
    return audio, sr

def trim_silence(audio, sr, top_db=30):
    trimmed, _ = librosa.effects.trim(audio, top_db=top_db)
    return trimmed

def normalize(audio):
    peak = np.max(np.abs(audio))
    return audio / peak if peak >= 1e-8 else audio

def pad_or_segment(audio, sr, target_duration, hop_duration):
    target_samples = int(target_duration * sr)
    hop_samples    = int(hop_duration * sr)
    n_samples      = len(audio)
    if n_samples <= target_samples:
        padded = np.zeros(target_samples, dtype=np.float32)
        padded[:n_samples] = audio
        yield padded
    else:
        start = 0
        while start + target_samples <= n_samples:
            yield audio[start: start + target_samples].astype(np.float32)
            start += hop_samples
        remaining = audio[start:]
        if len(remaining) >= int(0.5 * sr):
            padded = np.zeros(target_samples, dtype=np.float32)
            padded[:len(remaining)] = remaining
            yield padded

def waveform_hash(audio, dtype="float32"):
    return hashlib.sha256(audio.astype(dtype).tobytes()).hexdigest()

def passes_quality_filter(audio, sr, min_duration, max_silence_ratio, silence_threshold):
    if len(audio) / sr < min_duration:
        return False
    if _silence_ratio(audio, silence_threshold) > max_silence_ratio:
        return False
    return True

def standardize_file(input_path, output_path, target_sr, target_duration,
                     hop_duration, min_duration, max_silence_ratio,
                     silence_threshold, bit_depth="PCM_16"):
    try:
        audio, sr = load_and_resample(input_path, target_sr)
    except Exception:
        return []
    audio = trim_silence(audio, sr)
    audio = normalize(audio)
    if not passes_quality_filter(audio, sr, min_duration, max_silence_ratio, silence_threshold):
        return []
    results = []
    base_stem = Path(output_path).stem
    out_dir   = Path(output_path).parent
    out_dir.mkdir(parents=True, exist_ok=True)
    for i, chunk in enumerate(pad_or_segment(audio, sr, target_duration, hop_duration)):
        chunk_path = str(output_path) if i == 0 else str(out_dir / f"{base_stem}_seg{i:03d}.wav")
        sf.write(chunk_path, chunk, sr, subtype=bit_depth)
        results.append({
            "output_path":   chunk_path,
            "waveform_hash": waveform_hash(chunk),
            "duration":      len(chunk) / sr,
            "sr":            sr,
            "chunk_index":   i,
        })
    return results
''')

# ── pipeline/deduplicate.py ───────────────────────────────────────────────
open('/content/pipeline/audio_pipeline/pipeline/deduplicate.py', 'w').write('''
import numpy as np
import pandas as pd

DATASET_PRIORITY    = {"desed": 0, "sins": 1, "mimii": 2, "datasec": 3}
ANNOTATION_PRIORITY = {"strong": 0, "weak": 1, "none": 2}

def _priority_score(row):
    dp = DATASET_PRIORITY.get(str(row.get("source_dataset", "")), 99)
    ap = ANNOTATION_PRIORITY.get(str(row.get("annotation_type", "none")), 2)
    return dp * 10 + ap

def deduplicate_exact(df):
    df = df.copy()
    df["_priority"] = df.apply(_priority_score, axis=1)
    df = df.sort_values("_priority")
    mask = ~df["waveform_hash"].duplicated(keep="first")
    print(f"  [Exact dedup] Removed {(~mask).sum():,} duplicates -> {mask.sum():,} remaining")
    return df[mask].drop(columns=["_priority"]).reset_index(drop=True)
''')

# ── pipeline/split.py ─────────────────────────────────────────────────────
open('/content/pipeline/audio_pipeline/pipeline/split.py', 'w').write('''
import numpy as np
import pandas as pd
from pathlib import Path

def stratified_split(df, ratios, seed):
    rng = np.random.default_rng(seed)
    if "recording_id" not in df.columns:
        df = df.copy()
        df["recording_id"] = df["clip_id"]
    stratum_key = df["unified_label"].astype(str) + "||" + df["source_dataset"].astype(str)
    split_col   = pd.Series(["train"] * len(df), index=df.index)
    warnings_list = []
    for stratum in stratum_key.unique():
        mask   = (stratum_key == stratum)
        idx    = df.index[mask].to_numpy()
        groups = df.loc[idx, "recording_id"].values
        unique_groups = np.unique(groups)
        if len(unique_groups) < 3:
            warnings_list.append(f"{stratum} has <3 recordings")
            continue
        n_test  = max(1, int(round(len(unique_groups) * ratios["test"])))
        n_val   = max(1, int(round(len(unique_groups) * ratios["val"])))
        n_train = max(1, len(unique_groups) - n_val - n_test)
        perm     = rng.permutation(len(unique_groups))
        shuffled = unique_groups[perm]
        val_g    = set(shuffled[n_train: n_train + n_val])
        test_g   = set(shuffled[n_train + n_val:])
        for ci, grp in zip(idx, groups):
            if grp in val_g:   split_col.at[ci] = "val"
            elif grp in test_g: split_col.at[ci] = "test"
    df = df.copy()
    df["split"] = split_col
    print("\\n  Split distribution:")
    for s in ["train", "val", "test"]:
        n = (df["split"] == s).sum()
        print(f"    {s}: {n:,} clips ({100*n/len(df):.1f}%)")
    return df, warnings_list

def save_split_csvs(df, output_dir):
    meta_dir = Path(output_dir) / "metadata"
    meta_dir.mkdir(parents=True, exist_ok=True)
    df.to_csv(meta_dir / "master.csv", index=False)
    print(f"  Saved master.csv ({len(df):,} rows)")
    for split in ["train", "val", "test"]:
        sub = df[df["split"] == split]
        sub.to_csv(meta_dir / f"{split}.csv", index=False)
        print(f"  Saved {split}.csv ({len(sub):,} rows)")
''')

# ── pipeline/export_stats.py ──────────────────────────────────────────────
open('/content/pipeline/audio_pipeline/pipeline/export_stats.py', 'w').write('''
import json
from pathlib import Path
import sys
sys.path.insert(0, "/content/pipeline/audio_pipeline")
from configs.ontology import CLASS_TO_IDX, UNIFIED_CLASSES

def export_all(df, warnings, output_dir):
    meta_dir = Path(output_dir) / "metadata"
    meta_dir.mkdir(parents=True, exist_ok=True)
    labels = sorted(df["unified_label"].unique().tolist())
    label_to_idx = {l: i for i, l in enumerate(labels)}
    class_map = {
        "label_to_idx": label_to_idx,
        "idx_to_label": {str(v): k for k, v in label_to_idx.items()},
        "n_classes": len(label_to_idx),
    }
    with open(meta_dir / "class_map.json", "w") as f:
        json.dump(class_map, f, indent=2)
    stats = {
        "total_clips":      int(len(df)),
        "total_duration_h": round(df["duration"].sum() / 3600, 2),
        "n_classes":        int(df["unified_label"].nunique()),
        "n_datasets":       int(df["source_dataset"].nunique()),
        "splits":           {},
        "per_class":        {},
        "per_dataset":      {},
        "warnings":         warnings,
    }
    for split in ["train", "val", "test"]:
        sub = df[df["split"] == split]
        stats["splits"][split] = {
            "clips":      int(len(sub)),
            "duration_h": round(sub["duration"].sum() / 3600, 2),
        }
    for label, group in df.groupby("unified_label"):
        stats["per_class"][label] = {
            "total": int(len(group)),
            "train": int((group["split"] == "train").sum()),
            "val":   int((group["split"] == "val").sum()),
            "test":  int((group["split"] == "test").sum()),
        }
    for ds, group in df.groupby("source_dataset"):
        stats["per_dataset"][ds] = {
            "clips":      int(len(group)),
            "duration_h": round(group["duration"].sum() / 3600, 2),
        }
    with open(meta_dir / "stats.json", "w") as f:
        json.dump(stats, f, indent=2)
    sep = "=" * 52
    print(f"\\n{sep}\\n  UNIFIED DATASET REPORT\\n{sep}")
    print(f"  Total clips    : {stats[\'total_clips\']:,}")
    print(f"  Total hours    : {stats[\'total_duration_h\']} h")
    print(f"  Classes        : {stats[\'n_classes\']}")
    print(f"  Source datasets: {stats[\'n_datasets\']}")
    for split, s in stats["splits"].items():
        print(f"  {split}: {s[\'clips\']:,} clips ({s[\'duration_h\']} h)")
    for ds, s in stats["per_dataset"].items():
        print(f"  {ds}: {s[\'clips\']:,} clips ({s[\'duration_h\']} h)")
    print(sep)
''')

# ── pipeline/ingest.py ────────────────────────────────────────────────────
open('/content/pipeline/audio_pipeline/pipeline/ingest.py', 'w').write('''
import os
from pathlib import Path
import pandas as pd

def ingest_desed(root):
    records = []
    for tsv_path in root.rglob("*.tsv"):
        try:
            df = pd.read_csv(tsv_path, sep="\\t")
            has_strong = "onset" in df.columns and "offset" in df.columns
            for _, row in df.iterrows():
                fname = str(row.get("filename", ""))
                audio_path = root / fname
                if not audio_path.exists():
                    audio_path = root / "audio" / fname
                lbl = str(row.get("event_label", row.get("event_labels", "unknown")))
                for l in lbl.split(","):
                    records.append({
                        "input_path": str(audio_path),
                        "raw_label": l.strip(),
                        "annotation_type": "strong" if has_strong else "weak",
                        "onset": float(row["onset"]) if has_strong else None,
                        "offset": float(row["offset"]) if has_strong else None,
                        "recording_id": Path(fname).stem,
                        "source_dataset": "desed",
                        "license": "CC BY 4.0",
                        "snr_db": None,
                    })
        except Exception as e:
            pass
    print(f"  [DESED]   {len(records):,} entries")
    return records

def ingest_sins(root):
    records = []
    for meta_file in root.rglob("*.txt"):
        try:
            with open(meta_file) as f:
                for line in f:
                    parts = line.strip().split("\\t")
                    if len(parts) < 2:
                        continue
                    fname = parts[0]
                    label = parts[1] if len(parts) > 1 else "other"
                    audio_path = root / fname
                    if not audio_path.exists():
                        for candidate in root.rglob(Path(fname).name):
                            audio_path = candidate
                            break
                    records.append({
                        "input_path": str(audio_path),
                        "raw_label": label.strip(),
                        "annotation_type": "weak",
                        "onset": None, "offset": None,
                        "recording_id": Path(fname).stem,
                        "source_dataset": "sins",
                        "license": "CC BY 4.0",
                        "snr_db": None,
                    })
        except Exception:
            pass
    if not records:
        for wav in root.rglob("*.wav"):
            parts = wav.parts
            label = "other"
            for part in parts:
                if part in ["cooking", "watching_tv", "computer_work", "showering",
                            "toilet", "vacuum_cleaning", "eating",
                            "personal_hygiene", "absence", "other"]:
                    label = part
                    break
            records.append({
                "input_path": str(wav),
                "raw_label": label,
                "annotation_type": "weak",
                "onset": None, "offset": None,
                "recording_id": wav.stem,
                "source_dataset": "sins",
                "license": "CC BY 4.0",
                "snr_db": None,
            })
    print(f"  [SINS]    {len(records):,} entries")
    return records

def ingest_mimii(root):
    records = []
    for wav in root.rglob("*.wav"):
        parts = wav.parts
        machine = "unknown"
        condition = "normal"
        for part in parts:
            if part in ["fan", "pump", "slider", "valve"]:
                machine = part
            if part in ["normal", "abnormal", "anomaly"]:
                condition = "normal" if part == "normal" else "anomaly"
        records.append({
            "input_path": str(wav),
            "raw_label": f"{machine}/{condition}",
            "annotation_type": "weak",
            "onset": None, "offset": None,
            "recording_id": wav.stem,
            "source_dataset": "mimii",
            "license": "CC BY-SA 4.0",
            "snr_db": None,
        })
    print(f"  [MIMII]   {len(records):,} entries")
    return records

def ingest_datasec(root):
    records = []
    for wav in root.rglob("*.wav"):
        label = wav.parent.name
        records.append({
            "input_path": str(wav),
            "raw_label": label,
            "annotation_type": "weak",
            "onset": None, "offset": None,
            "recording_id": wav.stem,
            "source_dataset": "datasec",
            "license": "CC BY 4.0",
            "snr_db": None,
        })
    print(f"  [DataSEC] {len(records):,} entries")
    return records

def scan_dataset(name, root_dir, allowed_fsd_licenses=None):
    root = Path(root_dir)
    if not root.exists():
        print(f"  [SKIP] {name}: not found at {root}")
        return []
    dispatch = {
        "desed":   lambda: ingest_desed(root),
        "sins":    lambda: ingest_sins(root),
        "mimii":   lambda: ingest_mimii(root),
        "datasec": lambda: ingest_datasec(root),
    }
    fn = dispatch.get(name)
    if fn is None:
        print(f"  [SKIP] No adapter for {name}")
        return []
    return fn()
''')

# ── build_dataset.py ──────────────────────────────────────────────────────
open('/content/pipeline/audio_pipeline/build_dataset.py', 'w').write('''
import os, sys, uuid, time, argparse
from concurrent.futures import ProcessPoolExecutor, as_completed
from pathlib import Path
import pandas as pd
from tqdm import tqdm

sys.path.insert(0, "/content/pipeline/audio_pipeline")

from configs.settings  import (RAW_DATA_DIR, OUTPUT_DIR, TARGET_SR,
                                TARGET_DURATION, HOP_DURATION, N_FFT,
                                HOP_LENGTH, N_MELS, SPLIT_RATIOS, RANDOM_SEED,
                                MIN_CLIP_DURATION, MAX_SILENCE_RATIO,
                                SILENCE_THRESHOLD, FSD50K_ALLOWED_LICENCES,
                                DATASET_DIRS, AUDIO_FORMAT, BIT_DEPTH)
from configs.ontology  import LABEL_MAP
from pipeline.ingest   import scan_dataset
from pipeline.standardize import standardize_file
from pipeline.split    import stratified_split, save_split_csvs
from pipeline.export_stats import export_all

def parse_args():
    p = argparse.ArgumentParser()
    p.add_argument("--dry-run",    action="store_true")
    p.add_argument("--skip-dedup", action="store_true")
    p.add_argument("--dataset",    type=str, default=None)
    p.add_argument("--workers",    type=int, default=2)
    return p.parse_args()

def step_ingest(target_dataset):
    print(f"\\nINGEST & LABEL MAPPING")
    all_records = []
    datasets = DATASET_DIRS if target_dataset is None else {target_dataset: DATASET_DIRS[target_dataset]}
    for ds_name, ds_folder in datasets.items():
        raw_path = os.path.join(RAW_DATA_DIR, ds_folder)
        records  = scan_dataset(ds_name, raw_path, FSD50K_ALLOWED_LICENCES)
        mapped = skipped = 0
        for r in records:
            unified = LABEL_MAP.get(r["raw_label"]) or \
                      LABEL_MAP.get(r["raw_label"].strip().replace(" ", "_"))
            if unified:
                r["unified_label"] = unified
                mapped += 1
            else:
                r["unified_label"] = None
                skipped += 1
        records = [r for r in records if r["unified_label"] is not None]
        print(f"  {ds_name}: {mapped:,} mapped | {skipped:,} skipped")
        all_records.extend(records)
    print(f"  Total: {len(all_records):,} entries")
    return all_records

def _worker(args):
    record, output_base = args
    input_path = record["input_path"]
    if not os.path.exists(input_path):
        return []
    label    = record["unified_label"].replace("/", "_")
    stem     = Path(input_path).stem
    out_path = os.path.join(output_base, record["source_dataset"], label, f"{stem}.wav")
    os.makedirs(os.path.dirname(out_path), exist_ok=True)
    chunks = standardize_file(
        input_path, out_path, TARGET_SR, TARGET_DURATION, HOP_DURATION,
        MIN_CLIP_DURATION, MAX_SILENCE_RATIO, SILENCE_THRESHOLD, BIT_DEPTH)
    results = []
    for chunk in chunks:
        r = dict(record)
        r.update(chunk)
        r["clip_id"] = str(uuid.uuid4())
        results.append(r)
    return results

def step_standardize(records, workers):
    print(f"\\nSTANDARDIZE AUDIO")
    unique_inputs = {}
    for r in records:
        key = (r["input_path"], r["unified_label"])
        if key not in unique_inputs:
            unique_inputs[key] = r
    audio_dir = os.path.join(OUTPUT_DIR, "audio", "all")
    args_list = [(r, audio_dir) for r in unique_inputs.values()]
    print(f"  {len(args_list):,} files, {workers} workers...")
    all_chunks = []
    with ProcessPoolExecutor(max_workers=workers) as ex:
        futs = {ex.submit(_worker, a): a for a in args_list}
        for fut in tqdm(as_completed(futs), total=len(futs), ncols=70):
            try:
                all_chunks.extend(fut.result())
            except Exception:
                pass
    df = pd.DataFrame(all_chunks)
    print(f"  Produced {len(df):,} clips")
    return df

def step_split(df):
    print(f"\\nSTRATIFIED SPLIT")
    return stratified_split(df, SPLIT_RATIOS, RANDOM_SEED)

def step_reorganize(df):
    import shutil
    print(f"\\nREORGANIZING FILES")
    new_paths = []
    for _, row in tqdm(df.iterrows(), total=len(df), ncols=70):
        src = row["output_path"]
        if not os.path.exists(str(src)):
            new_paths.append(src)
            continue
        label    = str(row["unified_label"]).replace("/", "_")
        dest_dir = Path(OUTPUT_DIR) / "audio" / row["split"] / label
        dest_dir.mkdir(parents=True, exist_ok=True)
        dest = dest_dir / Path(src).name
        if not dest.exists():
            shutil.move(src, dest)
        new_paths.append(str(dest))
    df = df.copy()
    df["output_path"] = new_paths
    shutil.rmtree(Path(OUTPUT_DIR) / "audio" / "all", ignore_errors=True)
    return df

def main():
    args = parse_args()
    t0   = time.time()

    master_csv  = os.path.join(OUTPUT_DIR, "metadata", "master.csv")
    existing_df = pd.read_csv(master_csv) if os.path.exists(master_csv) else pd.DataFrame()

    records = step_ingest(args.dataset)
    if not records:
        print("No records found.")
        return

    df = step_standardize(records, args.workers)
    if df.empty:
        print("No clips survived.")
        return

    if not existing_df.empty:
        df = pd.concat([existing_df, df], ignore_index=True)
        print(f"  Combined total: {len(df):,} clips")

    df, warnings = step_split(df)
    df = step_reorganize(df)
    save_split_csvs(df, OUTPUT_DIR)
    export_all(df, warnings, OUTPUT_DIR)
    print(f"\\n  Done in {(time.time()-t0)/60:.1f} minutes.")

if __name__ == "__main__":
    main()
''')

print('All pipeline files written.')

# ── Verify ────────────────────────────────────────────────────────────────
required = ['configs/ontology.py', 'configs/settings.py',
            'pipeline/standardize.py', 'pipeline/deduplicate.py',
            'pipeline/split.py', 'pipeline/export_stats.py',
            'pipeline/ingest.py', 'build_dataset.py']

base = '/content/pipeline/audio_pipeline'
all_ok = True
for f in required:
    exists = os.path.exists(os.path.join(base, f))
    print(f"  {'OK  ' if exists else 'MISS'} {f}")
    if not exists:
        all_ok = False

if all_ok:
    print('\nAll files ready. Now run the processing cell.')
else:
    print('\nSomething missing — re-run this cell.')
```

```python
import os, sys, uuid, numpy as np, pandas as pd
from pathlib import Path
sys.path.insert(0, '/content/pipeline/audio_pipeline')

# ── Fix 1: DESED — find actual structure and ingest directly ──────────────
print('=== Fixing DESED ===')

# Find all TSV files in DESED
tsv_files = list(Path('/content/raw_data/DESED').rglob('*.tsv'))
print(f'Found {len(tsv_files)} TSV files:')
for t in tsv_files[:10]:
    print(f'  {t}')

records = []
for tsv_path in tsv_files:
    try:
        df_tsv = pd.read_csv(tsv_path, sep='\t')
        print(f'  {tsv_path.name}: columns = {list(df_tsv.columns)}')
        has_strong = 'onset' in df_tsv.columns and 'offset' in df_tsv.columns
        for _, row in df_tsv.iterrows():
            fname = str(row.get('filename', row.get('filename', '')))
            # Try multiple path locations
            candidates = [
                Path('/content/raw_data/DESED') / fname,
                Path('/content/raw_data/DESED') / 'audio' / fname,
                Path('/content/raw_data/DESED') / Path(fname).name,
            ]
            audio_path = None
            for c in candidates:
                if c.exists():
                    audio_path = c
                    break
            if audio_path is None:
                continue
            lbl = str(row.get('event_label', row.get('event_labels', 'unknown')))
            for l in lbl.split(','):
                l = l.strip()
                if l:
                    records.append({
                        'input_path': str(audio_path),
                        'raw_label': l,
                        'annotation_type': 'strong' if has_strong else 'weak',
                        'onset': float(row['onset']) if has_strong else None,
                        'offset': float(row['offset']) if has_strong else None,
                        'recording_id': Path(fname).stem,
                        'source_dataset': 'desed',
                        'license': 'CC BY 4.0',
                        'snr_db': None,
                    })
    except Exception as e:
        print(f'  Error reading {tsv_path.name}: {e}')

print(f'DESED records found: {len(records)}')

# If no TSV found, scan wav files directly by folder name
if len(records) == 0:
    print('No TSV annotations found. Scanning wav files by folder name...')
    label_folders = ['Alarm_bell_ringing', 'Blender', 'Cat', 'Dog', 'Dishes',
                     'Electric_shaver_toothbrush', 'Frying', 'Running_water',
                     'Speech', 'Vacuum_cleaner']
    for wav in Path('/content/raw_data/DESED').rglob('*.wav'):
        # Try to get label from parent folder name
        label = wav.parent.name
        if label not in label_folders:
            # Try grandparent
            label = wav.parent.parent.name
        records.append({
            'input_path': str(wav),
            'raw_label': label,
            'annotation_type': 'weak',
            'onset': None, 'offset': None,
            'recording_id': wav.stem,
            'source_dataset': 'desed',
            'license': 'CC BY 4.0',
            'snr_db': None,
        })
    print(f'DESED wav scan found: {len(records)} files')

# Show sample of what we found
if records:
    sample_labels = set(r['raw_label'] for r in records[:50])
    print(f'Sample labels: {sample_labels}')

# ── Fix 2: DataSEC split crash — fix mixed types then re-split ────────────
print('\n=== Fixing DataSEC split ===')

from configs.ontology import LABEL_MAP
from pipeline.standardize import standardize_file
from configs.settings import (TARGET_SR, TARGET_DURATION, HOP_DURATION,
                               MIN_CLIP_DURATION, MAX_SILENCE_RATIO,
                               SILENCE_THRESHOLD, BIT_DEPTH, OUTPUT_DIR)

# Load current master.csv (has SINS + MIMII already processed)
master_path = '/content/unified_dataset/metadata/master.csv'
existing_df = pd.read_csv(master_path, low_memory=False)
print(f'Existing clips (SINS + MIMII): {len(existing_df):,}')

# Fix recording_id type — convert everything to string
existing_df['recording_id'] = existing_df['recording_id'].astype(str)
existing_df['clip_id'] = existing_df['clip_id'].astype(str)
print('Fixed mixed types in existing data.')

# Process DESED records
print('\nProcessing DESED...')
from concurrent.futures import ProcessPoolExecutor, as_completed
from tqdm import tqdm

def process_record(args):
    record, output_base = args
    input_path = record['input_path']
    if not os.path.exists(input_path):
        return []
    unified = LABEL_MAP.get(record['raw_label']) or \
              LABEL_MAP.get(record['raw_label'].strip().replace(' ', '_'))
    if not unified:
        return []
    record['unified_label'] = unified
    label    = unified.replace('/', '_')
    stem     = Path(input_path).stem
    out_path = os.path.join(output_base, 'desed', label, f'{stem}.wav')
    os.makedirs(os.path.dirname(out_path), exist_ok=True)
    chunks = standardize_file(
        input_path, out_path, TARGET_SR, TARGET_DURATION, HOP_DURATION,
        MIN_CLIP_DURATION, MAX_SILENCE_RATIO, SILENCE_THRESHOLD, BIT_DEPTH)
    results = []
    for chunk in chunks:
        r = dict(record)
        r.update(chunk)
        r['clip_id'] = str(uuid.uuid4())
        r['recording_id'] = str(r.get('recording_id', stem))
        results.append(r)
    return results

if records:
    unique_records = {}
    for r in records:
        key = (r['input_path'], r['raw_label'])
        if key not in unique_records:
            unique_records[key] = r

    audio_dir = os.path.join(OUTPUT_DIR, 'audio', 'all')
    args_list = [(r, audio_dir) for r in unique_records.values()]
    print(f'  Processing {len(args_list):,} DESED files...')

    desed_chunks = []
    with ProcessPoolExecutor(max_workers=2) as ex:
        futs = {ex.submit(process_record, a): a for a in args_list}
        for fut in tqdm(as_completed(futs), total=len(futs), ncols=70):
            try:
                desed_chunks.extend(fut.result())
            except Exception:
                pass
    print(f'  DESED produced: {len(desed_chunks):,} clips')
else:
    desed_chunks = []
    print('  No DESED records to process.')

# Process DataSEC records
print('\nProcessing DataSEC...')
datasec_records = []
for wav in Path('/content/raw_data/DataSEC_DataSED').rglob('*.wav'):
    label = wav.parent.name
    unified = LABEL_MAP.get(label)
    if unified:
        datasec_records.append({
            'input_path': str(wav),
            'raw_label': label,
            'unified_label': unified,
            'annotation_type': 'weak',
            'onset': None, 'offset': None,
            'recording_id': str(wav.stem),
            'source_dataset': 'datasec',
            'license': 'CC BY 4.0',
            'snr_db': None,
        })

print(f'  DataSEC: {len(datasec_records):,} mapped records')

datasec_args = [(r, os.path.join(OUTPUT_DIR, 'audio', 'all'))
                for r in datasec_records]
datasec_chunks = []
with ProcessPoolExecutor(max_workers=2) as ex:
    futs = {ex.submit(process_record, a): a for a in datasec_args}
    for fut in tqdm(as_completed(futs), total=len(futs), ncols=70):
        try:
            datasec_chunks.extend(fut.result())
        except Exception:
            pass
print(f'  DataSEC produced: {len(datasec_chunks):,} clips')

# ── Combine everything and run final split ────────────────────────────────
print('\n=== Combining all datasets ===')

new_df = pd.DataFrame(desed_chunks + datasec_chunks)
if not new_df.empty:
    new_df['recording_id'] = new_df['recording_id'].astype(str)
    new_df['clip_id']      = new_df['clip_id'].astype(str)

all_df = pd.concat([existing_df, new_df], ignore_index=True) if not new_df.empty else existing_df
all_df['recording_id'] = all_df['recording_id'].astype(str)
all_df['clip_id']      = all_df['clip_id'].astype(str)
print(f'Total clips: {len(all_df):,}')
print(f'Sources: {all_df["source_dataset"].unique().tolist()}')

# ── Deduplication ─────────────────────────────────────────────────────────
print('\n=== Deduplication ===')
PRIORITY = {'desed': 0, 'sins': 1, 'mimii': 2, 'datasec': 3}
all_df['_p'] = all_df.apply(
    lambda r: PRIORITY.get(str(r.get('source_dataset','')), 99)*10 +
              {'strong':0,'weak':1,'none':2}.get(str(r.get('annotation_type','none')),2),
    axis=1)
all_df = all_df.sort_values('_p')
mask = ~all_df['waveform_hash'].duplicated(keep='first')
print(f'  Removed {(~mask).sum():,} duplicates → {mask.sum():,} remaining')
all_df = all_df[mask].drop(columns=['_p']).reset_index(drop=True)

# ── Final split ───────────────────────────────────────────────────────────
print('\n=== Final split ===')
rng = np.random.default_rng(42)
stratum_key = all_df['unified_label'].astype(str) + '||' + all_df['source_dataset'].astype(str)
split_col   = pd.Series(['train'] * len(all_df), index=all_df.index)

for stratum in stratum_key.unique():
    mask   = (stratum_key == stratum)
    idx    = all_df.index[mask].to_numpy()
    groups = all_df.loc[idx, 'recording_id'].astype(str).values
    unique_groups = np.unique(groups)
    if len(unique_groups) < 3:
        continue
    n_test  = max(1, int(round(len(unique_groups) * 0.15)))
    n_val   = max(1, int(round(len(unique_groups) * 0.15)))
    n_train = max(1, len(unique_groups) - n_val - n_test)
    perm     = rng.permutation(len(unique_groups))
    shuffled = unique_groups[perm]
    val_g    = set(shuffled[n_train: n_train + n_val])
    test_g   = set(shuffled[n_train + n_val:])
    for ci, grp in zip(idx, groups):
        if grp in val_g:    split_col.at[ci] = 'val'
        elif grp in test_g: split_col.at[ci] = 'test'

all_df['split'] = split_col

for s in ['train', 'val', 'test']:
    n = (all_df['split'] == s).sum()
    print(f'  {s}: {n:,} clips ({100*n/len(all_df):.1f}%)')

# ── Reorganize files ──────────────────────────────────────────────────────
print('\n=== Reorganizing files ===')
import shutil
new_paths = []
for _, row in tqdm(all_df.iterrows(), total=len(all_df), ncols=70):
    src = str(row['output_path'])
    if not os.path.exists(src):
        new_paths.append(src)
        continue
    label    = str(row['unified_label']).replace('/', '_')
    dest_dir = Path(OUTPUT_DIR) / 'audio' / row['split'] / label
    dest_dir.mkdir(parents=True, exist_ok=True)
    dest = dest_dir / Path(src).name
    if not dest.exists():
        shutil.move(src, dest)
    new_paths.append(str(dest))
all_df['output_path'] = new_paths
shutil.rmtree(Path(OUTPUT_DIR) / 'audio' / 'all', ignore_errors=True)

# ── Save metadata ─────────────────────────────────────────────────────────
print('\n=== Saving metadata ===')
import json
meta_dir = Path(OUTPUT_DIR) / 'metadata'
meta_dir.mkdir(parents=True, exist_ok=True)

all_df.to_csv(meta_dir / 'master.csv', index=False)
for split in ['train', 'val', 'test']:
    sub = all_df[all_df['split'] == split]
    sub.to_csv(meta_dir / f'{split}.csv', index=False)

labels = sorted(all_df['unified_label'].unique().tolist())
class_map = {
    'label_to_idx': {l: i for i, l in enumerate(labels)},
    'idx_to_label': {str(i): l for i, l in enumerate(labels)},
    'n_classes': len(labels),
}
with open(meta_dir / 'class_map.json', 'w') as f:
    json.dump(class_map, f, indent=2)

stats = {
    'total_clips':      int(len(all_df)),
    'total_duration_h': round(all_df['duration'].sum() / 3600, 2),
    'n_classes':        int(all_df['unified_label'].nunique()),
    'n_datasets':       int(all_df['source_dataset'].nunique()),
    'splits':           {s: {'clips': int((all_df['split']==s).sum())} for s in ['train','val','test']},
    'per_dataset':      {ds: {'clips': int(len(g))} for ds, g in all_df.groupby('source_dataset')},
}
with open(meta_dir / 'stats.json', 'w') as f:
    json.dump(stats, f, indent=2)

# ── Final report ──────────────────────────────────────────────────────────
sep = '=' * 52
print(f'\n{sep}')
print('  FINAL UNIFIED DATASET')
print(sep)
print(f"  Total clips    : {stats['total_clips']:,}")
print(f"  Total hours    : {stats['total_duration_h']} h")
print(f"  Classes        : {stats['n_classes']}")
print(f"  Datasets       : {stats['n_datasets']}")
for s in ['train', 'val', 'test']:
    n = stats['splits'][s]['clips']
    print(f"  {s}: {n:,} clips")
for ds, s in stats['per_dataset'].items():
    print(f"  {ds}: {s['clips']:,} clips")
print(sep)
print('\nDone. Dataset is ready.')
```
